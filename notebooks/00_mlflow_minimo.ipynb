```python
import mlflow
import sklearn

print("Mlflow: ", mlflow.__version__)
print("Scikit-learn: ", sklearn.__version__)
```

```python
mlflow.set_experiment("mlflow-prueba-minima")
```

```python
with mlflow.start_run(run_name="prueba-01") as run:

    mlflow.log_param("modelo", "prueba")
    mlflow.log_param("random_state", 42)

    mlflow.log_metric("rmse", 0.75)
    mlflow.log_metric("r2", 0.82)

    print("Run ID: ", run.info.run_id)
```

```python
with mlflow.start_run(run_name="prueba-02") as run:

    mlflow.log_params({
        "modelo": "prueba",
        "random_state": 42,
        "n_estimators": 100
    })

    mlflow.log_metrics({
        "rmse": 0.61,
        "r2": 0.85
    })

    print("Run ID: ", run.info.run_id)
```
